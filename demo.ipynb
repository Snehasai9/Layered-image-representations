```python
import matplotlib.pyplot as plt
import glob
import os
from PIL import Image

# 3. Visualize generated RGBA layers
layer_files = sorted([f for f in glob.glob("outputs/colab_run/layer_*.png") if not ("albedo" in f or "shading" in f)])
fig, axes = plt.subplots(1, len(layer_files), figsize=(18, 4))
for ax, f in zip(axes, layer_files):
    img = Image.open(f)
    ax.imshow(img)
    ax.set_title(os.path.basename(f), fontsize=8)
    ax.axis('off')
plt.tight_layout()
plt.show()
```
